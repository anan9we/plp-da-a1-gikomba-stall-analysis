## Introduction

This notebook analyses 14 days of sales and stock records for Wanjiku's fresh-produce stall at Gikomba Market. The goal is to evaluate business performance, identify sales trends, classify daily performance, and determine which products need restocking.

```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]  # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2


# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

## Task 1: Sales Performance

In this task, I will calculate the total, average, highest and lowest sales over the 14-day period. I will also identify the day and week in which the highest and lowest sales occurred.

```python
total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)

highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

best_index = daily_sales.index(highest_sales)
worst_index = daily_sales.index(lowest_sales)

best_day = days[best_index]
worst_day = days[worst_index]

best_week = (best_index // 7) + 1
worst_week = (worst_index // 7) + 1

print("Total sales:", total_sales)
print("Average daily sales:", average_sales)
print("Highest sales:", highest_sales)
print("Best day:", best_day, "- Week", best_week)
print("Lowest sales:", lowest_sales)
print("Worst day:", worst_day, "- Week", worst_week)
```

## Task 2: Weekly Sales Growth

In this task, I will create a function called week_totals() to calculate the total sales for each week. I will then calculate the percentage growth or decline in sales from Week 1 to Week 2.

```python
def week_totals(sales):
    """Calculate and return the sales totals for two weeks."""
    week1 = sum(sales[:7])
    week2 = sum(sales[7:])
    return week1, week2


week1_total, week2_total = week_totals(daily_sales)

growth_pct = ((week2_total - week1_total) / week1_total) * 100

print(f"Week 1 total: KES {week1_total:,.2f}")
print(f"Week 2 total: KES {week2_total:,.2f}")
print(f"Week-on-week growth: {growth_pct:.2f}%")
```

## Task 3: Daily Performance Classification

In this task, I will classify each day's sales as Strong, Normal or Weak based on the overall average sales. I will then count the number of days in each category.

```python
def classify_day(amount, average):
    """Classify daily sales based on the overall average."""
    if amount > average * 1.2:
        return "Strong"
    elif amount < average * 0.8:
        return "Weak"
    else:
        return "Normal"


# Create a dictionary to count each category
performance_counts = {
    "Strong": 0,
    "Normal": 0,
    "Weak": 0
}

# Classify each day's sales
for day, amount in zip(days, daily_sales):
    category = classify_day(amount, average_sales)
    performance_counts[category] += 1
    print(f"{day}: KES {amount:,.2f} - {category}")

print("\nPerformance Summary:")
print(performance_counts)
```

## Task 4: Weekday vs Weekend Analysis

In this task, I will calculate the average daily sales for weekdays (Monday to Friday) and weekends (Saturday and Sunday). I will then compare the two averages to determine the percentage difference.

```python
weekday_sales = []
weekend_sales = []

for day, amount in zip(days, daily_sales):
    if day in ["Mon", "Tue", "Wed", "Thu", "Fri"]:
        weekday_sales.append(amount)
    else:
        weekend_sales.append(amount)

weekday_average = sum(weekday_sales) / len(weekday_sales)
weekend_average = sum(weekend_sales) / len(weekend_sales)

percentage_difference = (
    (weekend_average - weekday_average) / weekday_average
) * 100

print(f"Average weekday sales: KES {weekday_average:,.2f}")
print(f"Average weekend sales: KES {weekend_average:,.2f}")
print(f"Weekend sales difference: {percentage_difference:.2f}%")
```

## Task 5: Inventory and Gross Margins

In this task, I will identify products whose stock quantities are at or below their reorder levels. I will also calculate the gross margin percentage for each product using its cost and selling price.

```python
def reorder_products(items):
    """Return products whose stock is at or below their reorder level."""
    to_reorder = []

    for name, details in items.items():
        if details["qty"] <= details["reorder"]:
            to_reorder.append(name)

    return to_reorder


# Identify products that need restocking
reorder_list = reorder_products(stock)

print("Products to reorder:")
print(reorder_list)


# Calculate gross margin for each product
print("\nGross Margins:")

for name, details in stock.items():
    margin = ((details["price"] - details["cost"]) / details["price"]) * 100
    print(f"{name}: {margin:.2f}%")
```

## Task 6: Stall Performance Report

This section presents a summary of Wanjiku's business performance over the 14-day period, including sales, weekly growth, daily classifications, weekend performance and inventory status.

```python
print("=" * 45)
print("         STALL PERFORMANCE REPORT")
print("=" * 45)

print("\nSALES SUMMARY")
print("-" * 45)
print(f"Total sales:          KES {total_sales:,.2f}")
print(f"Average daily sales:  KES {average_sales:,.2f}")
print(f"Highest sales:        KES {highest_sales:,.2f}")
print(f"Lowest sales:         KES {lowest_sales:,.2f}")
print(f"Best day:             {best_day}, Week {best_week}")
print(f"Worst day:            {worst_day}, Week {worst_week}")

print("\nWEEKLY PERFORMANCE")
print("-" * 45)
print(f"Week 1 sales:         KES {week1_total:,.2f}")
print(f"Week 2 sales:         KES {week2_total:,.2f}")
print(f"Weekly growth:        {growth_pct:.2f}%")

print("\nDAILY CLASSIFICATION")
print("-" * 45)
for category, count in performance_counts.items():
    print(f"{category:<20} {count} days")

print("\nWEEKDAY VS WEEKEND")
print("-" * 45)
print(f"Weekday average:      KES {weekday_average:,.2f}")
print(f"Weekend average:     KES {weekend_average:,.2f}")
print(f"Weekend advantage:    {percentage_difference:.2f}%")

print("\nINVENTORY")
print("-" * 45)
print(f"Products to reorder:  {', '.join(reorder_list)}")

for name, details in stock.items():
    margin = ((details["price"] - details["cost"]) / details["price"]) * 100
    print(f"{name:<15} Gross margin: {margin:.2f}%")

print("=" * 45)
print("              END OF REPORT")
print("=" * 45)
```
